```python
# ============================================================
# EDGE SIMULATION — STEP 2
# RECREATE & SAVE INT8 SHUFFLENETV2
# ============================================================

import copy
import json
from pathlib import Path

import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/diffusion_project")

ORIGINAL_DIR = BASE_DIR / "results" / "shufflenetv2"
EDGE_DIR = BASE_DIR / "results" / "edge_simulation"

FP32_CHECKPOINT = ORIGINAL_DIR / "checkpoint.pth"
TEST_DIR = BASE_DIR / "dataset_final" / "test"

ARTIFACT_DIR = EDGE_DIR / "artifacts"

INT8_ARTIFACT = ARTIFACT_DIR / "shufflenetv2_int8_edge.pth"

# ------------------------------------------------------------
# SAFETY
# ------------------------------------------------------------

if not FP32_CHECKPOINT.exists():
    raise FileNotFoundError("STOP: FP32 checkpoint missing.")

if INT8_ARTIFACT.exists():
    raise FileExistsError(
        f"STOP: Edge INT8 artifact already exists:\n{INT8_ARTIFACT}"
    )

if EDGE_DIR.resolve() == ORIGINAL_DIR.resolve():
    raise RuntimeError("STOP: Edge directory equals original directory.")

print("=" * 60)
print("EDGE SIMULATION — INT8 ARTIFACT CREATION")
print("=" * 60)

print("\nOriginal checkpoint:")
print(FP32_CHECKPOINT)

print("\nNew INT8 artifact:")
print(INT8_ARTIFACT)

# ------------------------------------------------------------
# DEVICE
# ------------------------------------------------------------

device = torch.device("cpu")

print("\nQuantization device:", device)

# ------------------------------------------------------------
# DATA
# ------------------------------------------------------------

IMG_SIZE = 224

eval_tf = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

train_ds = datasets.ImageFolder(
    BASE_DIR / "dataset_final" / "train",
    transform=eval_tf
)

train_loader = DataLoader(
    train_ds,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

print("\nClasses:", train_ds.classes)
print("Calibration images:", len(train_ds))

# ------------------------------------------------------------
# BUILD EXACT SHUFFLENETV2 ARCHITECTURE
# ------------------------------------------------------------

from torchvision.models import shufflenet_v2_x1_0

model = shufflenet_v2_x1_0(
    weights=None,
    num_classes=2
)

# ------------------------------------------------------------
# LOAD ORIGINAL FP32 CHECKPOINT — READ ONLY
# ------------------------------------------------------------

raw_state_dict = torch.load(
    FP32_CHECKPOINT,
    map_location="cpu"
)

clean_state_dict = {
    k: v
    for k, v in raw_state_dict.items()
    if k not in {"total_ops", "total_params"}
    and not k.endswith(".total_ops")
    and not k.endswith(".total_params")
}

missing, unexpected = model.load_state_dict(
    clean_state_dict,
    strict=False
)

print("\nCheckpoint verification:")
print("Missing keys:", missing)
print("Unexpected keys:", unexpected)

if missing or unexpected:
    raise RuntimeError(
        "STOP: Checkpoint mismatch."
    )

model.eval()

print("\nOriginal FP32 checkpoint loaded successfully.")
print("Original checkpoint was NOT modified.")

# ------------------------------------------------------------
# QUANTIZATION ENGINE
# ------------------------------------------------------------

supported = torch.backends.quantized.supported_engines

print("\nSupported quantized engines:", supported)

if "fbgemm" in supported:
    engine = "fbgemm"
elif "qnnpack" in supported:
    engine = "qnnpack"
else:
    engine = supported[0]

torch.backends.quantized.engine = engine

print("Using quantization engine:", engine)

# ------------------------------------------------------------
# FX STATIC QUANTIZATION
# ------------------------------------------------------------

from torch.quantization.quantize_fx import (
    prepare_fx,
    convert_fx
)

from torch.ao.quantization import (
    get_default_qconfig_mapping
)

qconfig_mapping = get_default_qconfig_mapping(engine)

example_input = torch.randn(
    1, 3, IMG_SIZE, IMG_SIZE
)

print("\nPreparing FX graph for INT8...")

prepared_model = prepare_fx(
    copy.deepcopy(model),
    qconfig_mapping,
    example_input
)

# ------------------------------------------------------------
# CALIBRATION
# ------------------------------------------------------------

CALIBRATION_BATCHES = 10

print(
    f"Calibrating using {CALIBRATION_BATCHES} batches..."
)

with torch.no_grad():

    for i, (images, _) in enumerate(train_loader):

        if i >= CALIBRATION_BATCHES:
            break

        prepared_model(images)

        print(
            f"Calibration batch {i + 1}/{CALIBRATION_BATCHES}"
        )

# ------------------------------------------------------------
# CONVERT TO INT8
# ------------------------------------------------------------

print("\nConverting model to INT8...")

quantized_model = convert_fx(
    prepared_model
)

quantized_model.eval()

print("INT8 conversion complete.")

# ------------------------------------------------------------
# SAVE ONLY TO EDGE DIRECTORY
# ------------------------------------------------------------

torch.save(
    quantized_model.state_dict(),
    INT8_ARTIFACT
)

print("\nINT8 artifact saved:")
print(INT8_ARTIFACT)

# ------------------------------------------------------------
# SIZE
# ------------------------------------------------------------

size_mb = (
    INT8_ARTIFACT.stat().st_size
    / (1024 * 1024)
)

print(
    f"\nINT8 edge artifact size: {size_mb:.2f} MB"
)

# ------------------------------------------------------------
# SAVE METADATA
# ------------------------------------------------------------

metadata_path = (
    ARTIFACT_DIR / "int8_artifact_metadata.json"
)

metadata = {
    "model": "shufflenetv2",
    "quantization": "INT8 static post-training",
    "quantization_engine": engine,
    "calibration_batches": CALIBRATION_BATCHES,
    "input_size": "224x224",
    "num_classes": 2,
    "classes": train_ds.classes,
    "artifact_size_mb": size_mb,
    "source_checkpoint": str(FP32_CHECKPOINT),
    "artifact_path": str(INT8_ARTIFACT),
    "original_checkpoint_modified": False
}

with open(metadata_path, "w") as f:
    json.dump(metadata, f, indent=2)

# ------------------------------------------------------------
# FINAL SAFETY CHECK
# ------------------------------------------------------------

assert FP32_CHECKPOINT.exists()
assert INT8_ARTIFACT.exists()
assert ORIGINAL_DIR.resolve() != EDGE_DIR.resolve()

print("\n" + "=" * 60)
print("STEP 2 COMPLETE — SAFETY CHECK PASSED")
print("=" * 60)

print("\nOriginal checkpoint still exists:", FP32_CHECKPOINT.exists())
print("New INT8 artifact exists:", INT8_ARTIFACT.exists())
print("Original directory was NOT written to.")
print("\nNext step: simulated edge-device inference.")
```

```python
# ============================================================
# STEP 3 — SIMULATED EDGE-DEVICE INFERENCE
# CORRECTED: LOAD INT8 STATE DICT + RECONSTRUCT MODEL
# ============================================================

import json
import time
import os
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torchvision.models import shufflenet_v2_x1_0
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
)

# ============================================================
# PATHS
# ============================================================

BASE_DIR = Path(
    "/content/drive/MyDrive/diffusion_project"
)

ORIGINAL_DIR = (
    BASE_DIR / "results" / "shufflenetv2"
)

EDGE_DIR = (
    BASE_DIR / "results" / "edge_simulation"
)

ARTIFACT_DIR = (
    EDGE_DIR / "artifacts"
)

RESULT_DIR = (
    EDGE_DIR / "results"
)

INT8_ARTIFACT = (
    ARTIFACT_DIR / "shufflenetv2_int8_edge.pth"
)

TEST_DIR = (
    BASE_DIR / "dataset_final" / "test"
)

RESULT_FILE = (
    RESULT_DIR /
    "shufflenetv2_edge_inference_results.json"
)


# ============================================================
# SAFETY
# ============================================================

print("=" * 60)
print("SIMULATED EDGE DEVICE — SAFETY CHECK")
print("=" * 60)

if not ORIGINAL_DIR.exists():
    raise FileNotFoundError(
        f"STOP: Original directory missing:\n{ORIGINAL_DIR}"
    )

if not INT8_ARTIFACT.exists():
    raise FileNotFoundError(
        f"STOP: INT8 artifact missing:\n{INT8_ARTIFACT}"
    )

if not TEST_DIR.exists():
    raise FileNotFoundError(
        f"STOP: Test dataset missing:\n{TEST_DIR}"
    )

RESULT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if RESULT_FILE.exists():
    raise FileExistsError(
        f"STOP: Result already exists:\n{RESULT_FILE}\n"
        "Nothing will be overwritten."
    )

print("Original directory:", ORIGINAL_DIR)
print("INT8 artifact:", INT8_ARTIFACT)
print("Test dataset:", TEST_DIR)
print("Output directory:", RESULT_DIR)

print("\nOriginal files will ONLY be READ.")
print("Existing INT8 artifact will ONLY be READ.")
print("New results will ONLY be written to edge_simulation.")


# ============================================================
# EDGE DEVICE CONFIGURATION
# ============================================================

print("\n" + "=" * 60)
print("SIMULATED EDGE DEVICE CONFIGURATION")
print("=" * 60)

device = torch.device("cpu")

# Simulate a constrained CPU environment.
torch.set_num_threads(1)

NUM_THREADS = torch.get_num_threads()
BATCH_SIZE = 1
IMG_SIZE = 224

print("Device:", device)
print("CPU threads:", NUM_THREADS)
print("Batch size:", BATCH_SIZE)
print("Input:", f"{IMG_SIZE} x {IMG_SIZE}")


# ============================================================
# DATASET
# ============================================================

eval_transform = transforms.Compose([
    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],
        std=[
            0.229,
            0.224,
            0.225
        ]
    ),
])

test_ds = datasets.ImageFolder(
    TEST_DIR,
    transform=eval_transform
)

print("\nDataset:")
print("Classes:", test_ds.classes)
print("Mapping:", test_ds.class_to_idx)
print("Test images:", len(test_ds))


# ============================================================
# LOAD INT8 STATE DICTIONARY
# ============================================================

print("\n" + "=" * 60)
print("LOADING INT8 STATE DICTIONARY")
print("=" * 60)

int8_state = torch.load(
    INT8_ARTIFACT,
    map_location="cpu",
    weights_only=False
)

if not isinstance(
    int8_state,
    dict
):
    raise RuntimeError(
        "Unexpected INT8 artifact format."
    )

print(
    "Artifact type:",
    type(int8_state)
)

print(
    "Number of state-dict entries:",
    len(int8_state)
)


# ============================================================
# RECONSTRUCT THE SAME QUANTIZED SHUFFLENETV2 ARCHITECTURE
# ============================================================

print("\n" + "=" * 60)
print("RECONSTRUCTING INT8 SHUFFLENETV2")
print("=" * 60)

# Build the original FP32 architecture.
model = shufflenet_v2_x1_0(
    weights=None,
    num_classes=2
)

model.eval()


# ------------------------------------------------------------
# IMPORTANT:
# The saved artifact contains quantized parameters.
# Therefore we reproduce the same FX quantization procedure
# used to create the artifact before loading its state dict.
# ------------------------------------------------------------

from torch.ao.quantization.quantize_fx import (
    prepare_fx,
    convert_fx,
)

from torch.ao.quantization import (
    get_default_qconfig_mapping,
)

engine = "fbgemm"

if engine not in torch.backends.quantized.supported_engines:
    raise RuntimeError(
        f"Required quantization engine '{engine}' "
        f"is unavailable.\n"
        f"Available engines: "
        f"{torch.backends.quantized.supported_engines}"
    )

torch.backends.quantized.engine = engine

print("Quantization engine:", engine)


# Example input used for FX graph preparation.
example_input = torch.randn(
    1,
    3,
    IMG_SIZE,
    IMG_SIZE
)


# ------------------------------------------------------------
# Prepare quantized graph
# ------------------------------------------------------------

qconfig_mapping = (
    get_default_qconfig_mapping(
        engine
    )
)

print("Preparing FX graph...")

prepared_model = prepare_fx(
    model,
    qconfig_mapping,
    example_input
)


# ============================================================
# CALIBRATION
# ============================================================

print("\nCalibrating reconstructed graph...")

# Calibration uses the same training data and 10 batches
# used by the original quantization pipeline.

calibration_ds = datasets.ImageFolder(
    BASE_DIR / "dataset_final" / "train",
    transform=eval_transform
)

calibration_loader = torch.utils.data.DataLoader(
    calibration_ds,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

with torch.no_grad():

    for i, (images, _) in enumerate(
        calibration_loader
    ):

        if i >= 10:
            break

        prepared_model(images)

        print(
            f"Calibration batch {i + 1}/10"
        )


# ============================================================
# CONVERT TO INT8
# ============================================================

print("\nConverting reconstructed graph to INT8...")

quantized_model = convert_fx(
    prepared_model
)

quantized_model.eval()

print("INT8 architecture reconstructed.")


# ============================================================
# LOAD EXISTING INT8 STATE DICT
# ============================================================

print("\nLoading existing INT8 state dictionary...")

missing, unexpected = (
    quantized_model.load_state_dict(
        int8_state,
        strict=False
    )
)

print("Missing keys:", missing)
print("Unexpected keys:", unexpected)

if missing or unexpected:
    raise RuntimeError(
        "INT8 state dictionary does not match "
        "the reconstructed architecture."
    )

print(
    "Existing INT8 artifact loaded successfully."
)

print(
    "Original checkpoint was NOT modified."
)


# ============================================================
# MODEL SIZE
# ============================================================

artifact_size_mb = (
    INT8_ARTIFACT.stat().st_size
    / (1024 * 1024)
)

print(
    f"\nINT8 artifact size: "
    f"{artifact_size_mb:.2f} MB"
)


# ============================================================
# WARM-UP
# ============================================================

print("\n" + "=" * 60)
print("EDGE INFERENCE")
print("=" * 60)

warmup_image, _ = test_ds[0]

warmup_image = (
    warmup_image
    .unsqueeze(0)
    .to(device)
)

with torch.no_grad():

    for _ in range(10):
        quantized_model(
            warmup_image
        )

print("Warm-up complete.")


# ============================================================
# SINGLE-IMAGE INFERENCE
# ============================================================

all_preds = []
all_labels = []
latencies_ms = []

print(
    "Running batch-size-1 inference..."
)

with torch.no_grad():

    for index in range(
        len(test_ds)
    ):

        image, label = test_ds[index]

        image = (
            image
            .unsqueeze(0)
            .to(device)
        )

        start = time.perf_counter()

        output = quantized_model(
            image
        )

        elapsed = (
            time.perf_counter()
            - start
        ) * 1000.0

        prediction = (
            output
            .argmax(dim=1)
            .item()
        )

        latencies_ms.append(
            elapsed
        )

        all_preds.append(
            prediction
        )

        all_labels.append(
            label
        )

        if (
            (index + 1) % 100
            == 0
        ):
            print(
                f"Processed "
                f"{index + 1}/"
                f"{len(test_ds)}"
            )


# ============================================================
# CLASSIFICATION METRICS
# ============================================================

accuracy = accuracy_score(
    all_labels,
    all_preds
)

precision, recall, f1, _ = (
    precision_recall_fscore_support(
        all_labels,
        all_preds,
        average="binary",
        zero_division=0
    )
)

cm = confusion_matrix(
    all_labels,
    all_preds
).tolist()


# ============================================================
# LATENCY METRICS
# ============================================================

avg_latency_ms = float(
    np.mean(latencies_ms)
)

median_latency_ms = float(
    np.median(latencies_ms)
)

p95_latency_ms = float(
    np.percentile(
        latencies_ms,
        95
    )
)

throughput = (
    1000.0
    / avg_latency_ms
)


# ============================================================
# MEMORY
# ============================================================

try:

    import psutil

    process = psutil.Process(
        os.getpid()
    )

    memory_mb = (
        process
        .memory_info()
        .rss
        / (1024 * 1024)
    )

except Exception:

    memory_mb = None


# ============================================================
# SAVE RESULTS
# ============================================================

results = {

    "experiment":
        "simulated_edge_device_inference",

    "model":
        "shufflenetv2",

    "quantization":
        "INT8 static",

    "execution_device":
        "CPU",

    "simulation": {
        "cpu_threads":
            NUM_THREADS,

        "batch_size":
            BATCH_SIZE,

        "input_size":
            "224x224",

        "single_image_inference":
            True
    },

    "dataset": {

        "name":
            "GenImage",

        "test_directory":
            str(TEST_DIR),

        "num_images":
            len(test_ds),

        "classes":
            test_ds.classes,

        "class_mapping":
            test_ds.class_to_idx
    },

    "classification": {

        "accuracy":
            float(accuracy),

        "precision":
            float(precision),

        "recall":
            float(recall),

        "f1":
            float(f1),

        "confusion_matrix":
            cm
    },

    "latency": {

        "average_ms_per_image":
            avg_latency_ms,

        "median_ms_per_image":
            median_latency_ms,

        "p95_ms_per_image":
            p95_latency_ms
    },

    "throughput": {

        "images_per_second":
            float(throughput)
    },

    "resource_usage": {

        "edge_artifact_size_mb":
            float(artifact_size_mb),

        "process_memory_mb":
            (
                float(memory_mb)
                if memory_mb is not None
                else None
            )
    },

    "artifact":
        str(INT8_ARTIFACT),

    "safety": {

        "original_checkpoint_modified":
            False,

        "original_results_modified":
            False,

        "existing_int8_artifact_modified":
            False,

        "output_directory":
            str(RESULT_DIR)
    }
}


with open(
    RESULT_FILE,
    "w"
) as f:

    json.dump(
        results,
        f,
        indent=2
    )


# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n" + "=" * 60)
print("SIMULATED EDGE DEVICE — COMPLETE")
print("=" * 60)

print(
    f"\nAccuracy:        {accuracy:.4f}"
)

print(
    f"Precision:       {precision:.4f}"
)

print(
    f"Recall:          {recall:.4f}"
)

print(
    f"F1-score:        {f1:.4f}"
)

print(
    f"\nAverage latency: "
    f"{avg_latency_ms:.2f} ms/image"
)

print(
    f"Median latency:  "
    f"{median_latency_ms:.2f} ms/image"
)

print(
    f"P95 latency:     "
    f"{p95_latency_ms:.2f} ms/image"
)

print(
    f"Throughput:      "
    f"{throughput:.2f} images/sec"
)

print(
    f"INT8 model size: "
    f"{artifact_size_mb:.2f} MB"
)

if memory_mb is not None:

    print(
        f"Process memory:  "
        f"{memory_mb:.2f} MB"
    )

print("\nResults saved ONLY to:")
print(RESULT_FILE)

print(
    "\nOriginal checkpoint modified: FALSE"
)

print(
    "Original results modified:    FALSE"
)

print(
    "INT8 artifact modified:       FALSE"
)

print("=" * 60)
```

```python
from pathlib import Path
import json
import time
import os
import numpy as np
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torchvision.models import resnet50
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

# ============================================================
# SAFETY / PATHS
# ============================================================

BASE_DIR = Path("/content/drive/MyDrive/diffusion_project")

ORIGINAL_DIR = BASE_DIR / "results" / "resnet50"
CHECKPOINT = ORIGINAL_DIR / "checkpoint.pth"

TEST_DIR = BASE_DIR / "dataset_final" / "test"

EDGE_DIR = BASE_DIR / "results" / "edge_simulation"
OUTPUT_DIR = EDGE_DIR / "results"

OUTPUT_FILE = OUTPUT_DIR / "resnet50_edge_inference_results.json"

print("=" * 60)
print("RESNET50 — SIMULATED EDGE DEVICE")
print("=" * 60)

if not ORIGINAL_DIR.exists():
    raise FileNotFoundError(f"STOP: Missing directory:\n{ORIGINAL_DIR}")

if not CHECKPOINT.exists():
    raise FileNotFoundError(f"STOP: Missing checkpoint:\n{CHECKPOINT}")

if not TEST_DIR.exists():
    raise FileNotFoundError(f"STOP: Missing test dataset:\n{TEST_DIR}")

if not EDGE_DIR.exists():
    raise FileNotFoundError(f"STOP: Missing edge directory:\n{EDGE_DIR}")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

if OUTPUT_FILE.exists():
    raise FileExistsError(
        f"STOP: Result already exists:\n{OUTPUT_FILE}\n"
        "Nothing was overwritten."
    )

print("Original ResNet50 directory:", ORIGINAL_DIR)
print("Checkpoint exists:", CHECKPOINT.exists())
print("Test dataset exists:", TEST_DIR.exists())
print("Output:", OUTPUT_FILE)

print("\nOriginal files will ONLY be READ.")
print("New results will ONLY be written to edge_simulation.")

# ============================================================
# EDGE DEVICE CONFIGURATION
# ============================================================

DEVICE = torch.device("cpu")

# Simulated constrained CPU
torch.set_num_threads(1)
torch.set_num_interop_threads(1)

BATCH_SIZE = 1
IMG_SIZE = 224

print("\n" + "=" * 60)
print("SIMULATED EDGE CONFIGURATION")
print("=" * 60)

print("Device:", DEVICE)
print("CPU threads:", torch.get_num_threads())
print("Batch size:", BATCH_SIZE)
print("Input:", f"{IMG_SIZE} x {IMG_SIZE}")

# ============================================================
# DATASET
# ============================================================

transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

test_ds = datasets.ImageFolder(
    TEST_DIR,
    transform=transform
)

print("\nDataset:")
print("Classes:", test_ds.classes)
print("Mapping:", test_ds.class_to_idx)
print("Test images:", len(test_ds))

# ============================================================
# MODEL
# ============================================================

print("\n" + "=" * 60)
print("LOADING RESNET50")
print("=" * 60)

model = resnet50(
    weights=None,
    num_classes=2
)

checkpoint = torch.load(
    CHECKPOINT,
    map_location="cpu"
)

# Remove THOP profiling entries if present.
state_dict = {
    k: v
    for k, v in checkpoint.items()
    if k not in {"total_ops", "total_params"}
    and not k.endswith(".total_ops")
    and not k.endswith(".total_params")
}

missing, unexpected = model.load_state_dict(
    state_dict,
    strict=False
)

print("Missing keys:", missing)
print("Unexpected keys:", unexpected)

if missing or unexpected:
    raise RuntimeError(
        "STOP: ResNet50 checkpoint mismatch."
    )

model.eval()
model.to(DEVICE)

print("ResNet50 loaded successfully.")
print("Original checkpoint was NOT modified.")

# ============================================================
# MODEL SIZE
# ============================================================

TEMP_SIZE_FILE = OUTPUT_DIR / "resnet50_edge_size_tmp.pth"

torch.save(
    model.state_dict(),
    TEMP_SIZE_FILE
)

model_size_mb = (
    TEMP_SIZE_FILE.stat().st_size
    / (1024 * 1024)
)

TEMP_SIZE_FILE.unlink()

print("\nResNet50 model size:", f"{model_size_mb:.2f} MB")

# ============================================================
# SINGLE-IMAGE INFERENCE
# ============================================================

def predict_one(image_tensor):

    image_tensor = image_tensor.unsqueeze(0).to(DEVICE)

    start = time.perf_counter()

    with torch.no_grad():
        output = model(image_tensor)

    elapsed = (
        time.perf_counter() - start
    ) * 1000

    prediction = output.argmax(
        dim=1
    ).item()

    return prediction, elapsed


# ============================================================
# WARM-UP
# ============================================================

print("\n" + "=" * 60)
print("EDGE INFERENCE")
print("=" * 60)

warmup_image, _ = test_ds[0]

for _ in range(10):
    predict_one(warmup_image)

print("Warm-up complete.")

# ============================================================
# EVALUATION
# ============================================================

all_labels = []
all_preds = []
latencies = []

print("Running batch-size-1 inference...")

for index in range(len(test_ds)):

    image, label = test_ds[index]

    prediction, latency = predict_one(
        image
    )

    all_preds.append(prediction)
    all_labels.append(label)
    latencies.append(latency)

    if (index + 1) % 100 == 0:
        print(
            f"Processed {index + 1}/{len(test_ds)}"
        )

# ============================================================
# METRICS
# ============================================================

accuracy = accuracy_score(
    all_labels,
    all_preds
)

precision, recall, f1, _ = (
    precision_recall_fscore_support(
        all_labels,
        all_preds,
        average="binary",
        zero_division=0
    )
)

latencies_np = np.array(
    latencies,
    dtype=np.float64
)

average_latency = float(
    np.mean(latencies_np)
)

median_latency = float(
    np.median(latencies_np)
)

p95_latency = float(
    np.percentile(latencies_np, 95)
)

throughput = (
    1000.0 / average_latency
)

# ============================================================
# RESULTS
# ============================================================

results = {
    "experiment": "simulated_edge_device_inference",
    "model": "resnet50",
    "precision": "FP32",

    "device": "CPU",
    "cpu_threads": torch.get_num_threads(),
    "batch_size": 1,

    "input_size": "224x224",

    "dataset": "GenImage",
    "test_images": len(test_ds),
    "classes": test_ds.classes,

    "accuracy": float(accuracy),
    "precision_score": float(precision),
    "recall": float(recall),
    "f1": float(f1),

    "model_size_mb": float(model_size_mb),

    "average_latency_ms": average_latency,
    "median_latency_ms": median_latency,
    "p95_latency_ms": p95_latency,

    "throughput_images_per_sec": float(
        throughput
    ),

    "checkpoint": str(CHECKPOINT),

    "safety": {
        "original_checkpoint_modified": False,
        "original_results_modified": False,
        "new_files_written_only_to_edge_simulation": True
    }
}

# ============================================================
# SAVE
# ============================================================

with open(OUTPUT_FILE, "w") as f:
    json.dump(
        results,
        f,
        indent=2
    )

# ============================================================
# FINAL OUTPUT
# ============================================================

print("\n" + "=" * 60)
print("RESNET50 EDGE SIMULATION — COMPLETE")
print("=" * 60)

print(
    f"\nAccuracy:        {accuracy:.4f}"
)

print(
    f"Precision:       {precision:.4f}"
)

print(
    f"Recall:          {recall:.4f}"
)

print(
    f"F1-score:        {f1:.4f}"
)

print(
    f"\nModel size:      {model_size_mb:.2f} MB"
)

print(
    f"Average latency: {average_latency:.2f} ms/image"
)

print(
    f"Median latency:  {median_latency:.2f} ms/image"
)

print(
    f"P95 latency:     {p95_latency:.2f} ms/image"
)

print(
    f"Throughput:      {throughput:.2f} images/sec"
)

print("\nResults saved ONLY to:")
print(OUTPUT_FILE)

print("\nOriginal checkpoint modified: FALSE")
print("Original results modified:    FALSE")

print("=" * 60)
```

```python
# ============================================================
# RESNET50 INT8 — SIMULATED EDGE DEVICE
# ============================================================
# SAFETY:
# - Original ResNet50 checkpoint is READ ONLY
# - Original results_int8.json is READ ONLY
# - Nothing inside results/resnet50 will be modified
# - New files are written ONLY to edge_simulation
# ============================================================

import os
import json
import time
import copy
from pathlib import Path

import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)
import timm

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = Path("/content/drive/MyDrive/diffusion_project")

ORIGINAL_DIR = BASE / "results" / "resnet50"
EDGE_DIR = BASE / "results" / "edge_simulation"

CHECKPOINT = ORIGINAL_DIR / "checkpoint.pth"
INT8_RESULTS = ORIGINAL_DIR / "results_int8.json"

DATA_DIR = BASE / "dataset_final"

ARTIFACT_DIR = EDGE_DIR / "artifacts"
RESULT_DIR = EDGE_DIR / "results"

ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)

INT8_ARTIFACT = ARTIFACT_DIR / "resnet50_int8_edge.pth"
EDGE_RESULTS = RESULT_DIR / "resnet50_int8_edge_inference_results.json"

# ------------------------------------------------------------
# SAFETY CHECK
# ------------------------------------------------------------

print("=" * 60)
print("RESNET50 INT8 — EDGE SIMULATION SAFETY CHECK")
print("=" * 60)

print(f"\nOriginal directory:")
print(ORIGINAL_DIR)
print("Exists:", ORIGINAL_DIR.exists())

print("\nOriginal checkpoint:")
print(CHECKPOINT)
print("Exists:", CHECKPOINT.exists())

print("\nOriginal INT8 results:")
print(INT8_RESULTS)
print("Exists:", INT8_RESULTS.exists())

print("\nTest dataset:")
print(DATA_DIR / "test")
print("Exists:", (DATA_DIR / "test").exists())

print("\nEdge artifact:")
print(INT8_ARTIFACT)

print("\nEdge results:")
print(EDGE_RESULTS)

if not CHECKPOINT.exists():
    raise FileNotFoundError("ResNet50 checkpoint not found.")

if not INT8_RESULTS.exists():
    raise FileNotFoundError("ResNet50 INT8 results.json not found.")

if not (DATA_DIR / "train").exists():
    raise FileNotFoundError("Training dataset not found.")

if not (DATA_DIR / "test").exists():
    raise FileNotFoundError("Test dataset not found.")

print("\nOriginal ResNet50 files will ONLY be READ.")
print("New files will ONLY be written to edge_simulation.")

# ------------------------------------------------------------
# EDGE DEVICE CONFIGURATION
# ------------------------------------------------------------

DEVICE = torch.device("cpu")

# Simulated constrained CPU
torch.set_num_threads(1)

BATCH_SIZE = 1
IMG_SIZE = 224
CALIBRATION_BATCHES = 10

print("\n" + "=" * 60)
print("SIMULATED EDGE CONFIGURATION")
print("=" * 60)

print("Device: CPU")
print("CPU threads:", torch.get_num_threads())
print("Batch size:", BATCH_SIZE)
print("Input:", f"{IMG_SIZE} x {IMG_SIZE}")

# ------------------------------------------------------------
# DATASET
# ------------------------------------------------------------

transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

train_dataset = datasets.ImageFolder(
    DATA_DIR / "train",
    transform=transform
)

test_dataset = datasets.ImageFolder(
    DATA_DIR / "test",
    transform=transform
)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=1,
    shuffle=False,
    num_workers=0
)

class_names = test_dataset.classes

print("\nDataset:")
print("Classes:", class_names)
print("Mapping:", train_dataset.class_to_idx)
print("Calibration images:", len(train_dataset))
print("Test images:", len(test_dataset))

# ------------------------------------------------------------
# BUILD RESNET50
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("LOADING RESNET50 FP32 CHECKPOINT")
print("=" * 60)

model = timm.create_model(
    "resnet50.a1_in1k",
    pretrained=False,
    num_classes=2
)

raw_state_dict = torch.load(
    CHECKPOINT,
    map_location="cpu"
)

if isinstance(raw_state_dict, dict) and "state_dict" in raw_state_dict:
    raw_state_dict = raw_state_dict["state_dict"]

clean_state_dict = {
    k: v
    for k, v in raw_state_dict.items()
    if not k.endswith("total_ops")
    and not k.endswith("total_params")
}

missing, unexpected = model.load_state_dict(
    clean_state_dict,
    strict=False
)

print("Missing keys:", missing)
print("Unexpected keys:", unexpected)

if missing or unexpected:
    raise RuntimeError(
        "Checkpoint verification failed."
    )

model.eval()
model.to(DEVICE)

print("ResNet50 loaded successfully.")
print("Original checkpoint was NOT modified.")

# ------------------------------------------------------------
# QUANTIZATION ENGINE
# ------------------------------------------------------------

supported_engines = torch.backends.quantized.supported_engines

print("\nSupported quantized engines:")
print(supported_engines)

if "fbgemm" in supported_engines:
    ENGINE = "fbgemm"
elif "qnnpack" in supported_engines:
    ENGINE = "qnnpack"
else:
    ENGINE = supported_engines[0]

torch.backends.quantized.engine = ENGINE

print("Using quantization engine:", ENGINE)

# ------------------------------------------------------------
# FX STATIC INT8 QUANTIZATION
# ------------------------------------------------------------

from torch.ao.quantization.quantize_fx import (
    prepare_fx,
    convert_fx
)
from torch.ao.quantization import (
    get_default_qconfig_mapping
)

print("\n" + "=" * 60)
print("PREPARING RESNET50 FOR INT8")
print("=" * 60)

qconfig_mapping = get_default_qconfig_mapping(ENGINE)

example_input = torch.randn(
    1,
    3,
    IMG_SIZE,
    IMG_SIZE
)

quant_model = copy.deepcopy(model)

prepared_model = prepare_fx(
    quant_model,
    qconfig_mapping,
    example_input
)

# ------------------------------------------------------------
# CALIBRATION
# ------------------------------------------------------------

print("\nCalibrating using", CALIBRATION_BATCHES, "batches...")

prepared_model.eval()

with torch.no_grad():
    for i, (images, _) in enumerate(train_loader):

        if i >= CALIBRATION_BATCHES:
            break

        images = images.to(DEVICE)

        prepared_model(images)

        print(
            f"Calibration batch "
            f"{i + 1}/{CALIBRATION_BATCHES}"
        )

# ------------------------------------------------------------
# CONVERT TO INT8
# ------------------------------------------------------------

print("\nConverting ResNet50 to INT8...")

quantized_model = convert_fx(
    prepared_model
)

quantized_model.eval()

print("INT8 conversion complete.")

# ------------------------------------------------------------
# SAVE INT8 EDGE ARTIFACT
# ------------------------------------------------------------

print("\nSaving INT8 edge artifact...")

torch.save(
    quantized_model.state_dict(),
    INT8_ARTIFACT
)

artifact_size_mb = (
    INT8_ARTIFACT.stat().st_size
    / (1024 * 1024)
)

print("INT8 artifact saved:")
print(INT8_ARTIFACT)

print(
    f"INT8 artifact size: "
    f"{artifact_size_mb:.2f} MB"
)

# ------------------------------------------------------------
# EDGE INFERENCE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SIMULATED EDGE INFERENCE")
print("=" * 60)

# Warm-up
with torch.no_grad():
    for images, _ in list(test_loader)[:10]:
        images = images.to(DEVICE)
        quantized_model(images)

print("Warm-up complete.")

predictions = []
labels = []
latencies = []

print("Running batch-size-1 inference...")

with torch.no_grad():

    for idx, (images, target) in enumerate(test_loader):

        images = images.to(DEVICE)

        start = time.perf_counter()

        output = quantized_model(images)

        elapsed = (
            time.perf_counter() - start
        ) * 1000

        latency = elapsed / images.size(0)

        latencies.append(latency)

        pred = output.argmax(dim=1).item()

        predictions.append(pred)
        labels.append(target.item())

        if (idx + 1) % 100 == 0:
            print(
                f"Processed "
                f"{idx + 1}/{len(test_dataset)}"
            )

# ------------------------------------------------------------
# METRICS
# ------------------------------------------------------------

accuracy = accuracy_score(
    labels,
    predictions
)

precision, recall, f1, _ = (
    precision_recall_fscore_support(
        labels,
        predictions,
        average="binary",
        zero_division=0
    )
)

avg_latency = sum(latencies) / len(latencies)

sorted_latencies = sorted(latencies)

median_latency = sorted_latencies[
    len(sorted_latencies) // 2
]

p95_index = int(
    0.95 * len(sorted_latencies)
) - 1

p95_latency = sorted_latencies[
    max(0, p95_index)
]

throughput = (
    1000 / avg_latency
)

# ------------------------------------------------------------
# MEMORY
# ------------------------------------------------------------

process_memory_mb = None

try:
    import resource

    memory_kb = resource.getrusage(
        resource.RUSAGE_SELF
    ).ru_maxrss

    # Linux reports KB
    process_memory_mb = memory_kb / 1024

except Exception:
    pass

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

results = {
    "model": "resnet50",
    "quantization": "INT8 static post-training",
    "device": "CPU",
    "cpu_threads": 1,
    "batch_size": 1,
    "input_size": "224x224",

    "accuracy": accuracy,
    "precision": precision,
    "recall": recall,
    "f1_score": f1,

    "average_latency_ms": avg_latency,
    "median_latency_ms": median_latency,
    "p95_latency_ms": p95_latency,
    "throughput_images_per_second": throughput,

    "model_size_mb": artifact_size_mb,
    "process_memory_mb": process_memory_mb,

    "quantization_engine": ENGINE,
    "calibration_batches": CALIBRATION_BATCHES,

    "artifact_path": str(INT8_ARTIFACT),
    "test_images": len(test_dataset)
}

with open(
    EDGE_RESULTS,
    "w"
) as f:
    json.dump(
        results,
        f,
        indent=2
    )

# ------------------------------------------------------------
# FINAL OUTPUT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("RESNET50 INT8 EDGE SIMULATION — COMPLETE")
print("=" * 60)

print(f"\nAccuracy:        {accuracy:.4f}")
print(f"Precision:       {precision:.4f}")
print(f"Recall:          {recall:.4f}")
print(f"F1-score:        {f1:.4f}")

print(f"\nINT8 model size: {artifact_size_mb:.2f} MB")

print(
    f"Average latency: "
    f"{avg_latency:.2f} ms/image"
)

print(
    f"Median latency:  "
    f"{median_latency:.2f} ms/image"
)

print(
    f"P95 latency:     "
    f"{p95_latency:.2f} ms/image"
)

print(
    f"Throughput:      "
    f"{throughput:.2f} images/sec"
)

if process_memory_mb is not None:
    print(
        f"Process memory:  "
        f"{process_memory_mb:.2f} MB"
    )

print("\nResults saved ONLY to:")
print(EDGE_RESULTS)

print("\nOriginal checkpoint modified: FALSE")
print("Original results modified:    FALSE")
print("Original INT8 results modified: FALSE")

print("\n" + "=" * 60)
```

```python
import json
from pathlib import Path

# ============================================================
# FINAL INT8 EDGE COMPARISON — READ-ONLY
# ============================================================

BASE = Path("/content/drive/MyDrive/diffusion_project/results")

SHUFFLE_EDGE = BASE / "edge_simulation/results/shufflenetv2_edge_inference_results.json"
RESNET_EDGE = BASE / "edge_simulation/results/resnet50_int8_edge_inference_results.json"

SHUFFLE_INT8 = BASE / "shufflenetv2/results_int8.json"
RESNET_INT8 = BASE / "resnet50/results_int8.json"

OUTPUT_DIR = BASE / "edge_simulation/results"
OUTPUT_FILE = OUTPUT_DIR / "final_int8_edge_comparison.json"

print("=" * 60)
print("FINAL INT8 EDGE COMPARISON — READ-ONLY")
print("=" * 60)

# ------------------------------------------------------------
# SAFETY CHECK
# ------------------------------------------------------------

for path in [SHUFFLE_EDGE, RESNET_EDGE, SHUFFLE_INT8, RESNET_INT8]:
    print(f"{path}: {path.exists()}")
    if not path.exists():
        raise FileNotFoundError(f"Missing required file: {path}")

print("\nAll required files found.")
print("Original files will ONLY be READ.")
print("New comparison file will ONLY be written to edge_simulation.")

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

with open(SHUFFLE_EDGE) as f:
    shuffle_edge = json.load(f)

with open(RESNET_EDGE) as f:
    resnet_edge = json.load(f)

with open(SHUFFLE_INT8) as f:
    shuffle_orig = json.load(f)

with open(RESNET_INT8) as f:
    resnet_orig = json.load(f)

# ------------------------------------------------------------
# HELPER
# ------------------------------------------------------------

def get_value(data, possible_paths):
    """
    Try several possible locations for a value.
    """
    for path in possible_paths:
        current = data

        try:
            for key in path:
                current = current[key]

            if current is not None:
                return current

        except (KeyError, TypeError):
            continue

    return None


# ------------------------------------------------------------
# SHUFFLENETV2 EDGE
# ------------------------------------------------------------

shuffle = {
    "accuracy": get_value(shuffle_edge, [
        ("classification", "accuracy"),
        ("accuracy",),
    ]),

    "precision": get_value(shuffle_edge, [
        ("classification", "precision"),
        ("precision",),
    ]),

    "recall": get_value(shuffle_edge, [
        ("classification", "recall"),
        ("recall",),
    ]),

    "f1": get_value(shuffle_edge, [
        ("classification", "f1"),
        ("f1",),
    ]),

    "model_size_mb": get_value(shuffle_edge, [
        ("resource_usage", "edge_artifact_size_mb"),
        ("model_size_mb",),
    ]),

    "average_latency_ms": get_value(shuffle_edge, [
        ("latency", "average_ms_per_image"),
        ("average_latency_ms",),
    ]),

    "median_latency_ms": get_value(shuffle_edge, [
        ("latency", "median_ms_per_image"),
        ("median_latency_ms",),
    ]),

    "p95_latency_ms": get_value(shuffle_edge, [
        ("latency", "p95_ms_per_image"),
        ("p95_latency_ms",),
    ]),

    "throughput_images_per_sec": get_value(shuffle_edge, [
        ("throughput", "images_per_second"),
        ("throughput_images_per_sec",),
    ]),

    "process_memory_mb": get_value(shuffle_edge, [
        ("resource_usage", "process_memory_mb"),
        ("process_memory_mb",),
    ]),
}

# ------------------------------------------------------------
# RESNET50 EDGE
# ------------------------------------------------------------

resnet = {
    "accuracy": get_value(resnet_edge, [
        ("classification", "accuracy"),
        ("accuracy",),
    ]),

    "precision": get_value(resnet_edge, [
        ("classification", "precision"),
        ("precision",),
    ]),

    "recall": get_value(resnet_edge, [
        ("classification", "recall"),
        ("recall",),
    ]),

    "f1": get_value(resnet_edge, [
        ("classification", "f1"),
        ("f1",),
        ("f1_score",),
    ]),

    "model_size_mb": get_value(resnet_edge, [
        ("resource_usage", "edge_artifact_size_mb"),
        ("int8_model_size_mb",),
        ("model_size_mb",),
    ]),

    "average_latency_ms": get_value(resnet_edge, [
        ("latency", "average_ms_per_image"),
        ("average_latency_ms",),
    ]),

    "median_latency_ms": get_value(resnet_edge, [
        ("latency", "median_ms_per_image"),
        ("median_latency_ms",),
    ]),

    "p95_latency_ms": get_value(resnet_edge, [
        ("latency", "p95_ms_per_image"),
        ("p95_latency_ms",),
    ]),

    "throughput_images_per_sec": get_value(resnet_edge, [
    ("throughput", "images_per_second"),
    ("throughput_images_per_sec",),
    ("throughput_images_per_second",),
]),

    "process_memory_mb": get_value(resnet_edge, [
        ("resource_usage", "process_memory_mb"),
        ("process_memory_mb",),
    ]),
}

# ------------------------------------------------------------
# FALLBACK: USE ORIGINAL INT8 RESULTS IF EDGE JSON
# DOES NOT CONTAIN A FIELD
# ------------------------------------------------------------

fallback_shuffle = {
    "accuracy": shuffle_orig.get("test_accuracy"),
    "precision": shuffle_orig.get("test_precision"),
    "recall": shuffle_orig.get("test_recall"),
    "f1": shuffle_orig.get("test_f1"),
    "model_size_mb": shuffle_orig.get("model_size_mb"),
}

fallback_resnet = {
    "accuracy": resnet_orig.get("test_accuracy"),
    "precision": resnet_orig.get("test_precision"),
    "recall": resnet_orig.get("test_recall"),
    "f1": resnet_orig.get("test_f1"),
    "model_size_mb": resnet_orig.get("model_size_mb"),
}

for key in fallback_shuffle:
    if shuffle[key] is None:
        shuffle[key] = fallback_shuffle[key]

for key in fallback_resnet:
    if resnet[key] is None:
        resnet[key] = fallback_resnet[key]

# ------------------------------------------------------------
# VERIFY REQUIRED VALUES
# ------------------------------------------------------------

required = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "model_size_mb",
    "average_latency_ms",
    "median_latency_ms",
    "p95_latency_ms",
    "throughput_images_per_sec",
]

for model_name, data in [
    ("ShuffleNetV2", shuffle),
    ("ResNet50", resnet),
]:
    missing = [k for k in required if data[k] is None]

    if missing:
        print(f"\nWARNING: {model_name} missing:")
        for item in missing:
            print(" -", item)

        print("\nAvailable JSON keys:")
        source = shuffle_edge if model_name == "ShuffleNetV2" else resnet_edge
        print(json.dumps(source, indent=2))

        raise RuntimeError(
            f"{model_name} still has missing comparison values: {missing}"
        )

# ------------------------------------------------------------
# TRADE-OFF CALCULATIONS
# ------------------------------------------------------------

size_reduction = (
    (1 - shuffle["model_size_mb"] / resnet["model_size_mb"])
    * 100
)

accuracy_difference = shuffle["accuracy"] - resnet["accuracy"]
f1_difference = shuffle["f1"] - resnet["f1"]

latency_difference = (
    shuffle["average_latency_ms"]
    - resnet["average_latency_ms"]
)

latency_reduction = (
    (1 - shuffle["average_latency_ms"]
     / resnet["average_latency_ms"])
    * 100
)

throughput_difference = (
    shuffle["throughput_images_per_sec"]
    - resnet["throughput_images_per_sec"]
)

throughput_change = (
    (shuffle["throughput_images_per_sec"]
     / resnet["throughput_images_per_sec"] - 1)
    * 100
)

# ------------------------------------------------------------
# PRINT FINAL RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SHUFFLENETV2 INT8 — EDGE")
print("=" * 60)

for key, value in shuffle.items():
    if value is not None:
        print(f"{key}: {value}")

print("\n" + "=" * 60)
print("RESNET50 INT8 — EDGE")
print("=" * 60)

for key, value in resnet.items():
    if value is not None:
        print(f"{key}: {value}")

print("\n" + "=" * 60)
print("FINAL LIGHTWEIGHT vs HEAVYWEIGHT TRADE-OFF")
print("=" * 60)

print(f"Model size reduction: {size_reduction:.2f}%")
print(f"Accuracy difference:  {accuracy_difference:+.4f}")
print(f"F1 difference:        {f1_difference:+.4f}")
print(f"Latency difference:   {latency_difference:+.2f} ms")
print(f"Latency reduction:    {latency_reduction:.2f}%")
print(f"Throughput difference:{throughput_difference:+.2f} images/sec")
print(f"Throughput change:    {throughput_change:+.2f}%")

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

final_results = {
    "experiment": "final_int8_edge_device_comparison",

    "device": {
        "cpu_threads": 1,
        "batch_size": 1,
        "input_size": "224x224",
    },

    "models": {
        "shufflenetv2_int8": shuffle,
        "resnet50_int8": resnet,
    },

    "tradeoffs": {
        "model_size_reduction_percent": size_reduction,
        "accuracy_difference": accuracy_difference,
        "f1_difference": f1_difference,
        "latency_difference_ms": latency_difference,
        "latency_reduction_percent": latency_reduction,
        "throughput_difference_images_per_sec": throughput_difference,
        "throughput_change_percent": throughput_change,
    },

    "safety": {
        "original_files_modified": False,
        "output_location": str(OUTPUT_FILE),
    },
}

with open(OUTPUT_FILE, "w") as f:
    json.dump(final_results, f, indent=2)

print("\n" + "=" * 60)
print("FINAL COMPARISON SAVED")
print("=" * 60)
print(OUTPUT_FILE)

print("\nOriginal files were NOT modified.")
print("Edge simulation comparison complete.")
```
